## Objective

The objective of this sprint is to explore the dataset visually and statistically to discover hidden patterns, relationships, trends, and anomalies.

Exploratory Data Analysis helps us understand the data before feature engineering and model building.

The insights gained during EDA guide preprocessing decisions and improve the quality of machine learning models.

# What is Exploratory Data Analysis?

Exploratory Data Analysis (EDA) is the process of examining a dataset using statistics and visualizations before building machine learning models.

EDA helps answer questions such as:

- How is the target variable distributed?
- Which features influence the target?
- Are there outliers?
- Are there missing values?
- Are variables correlated?
- Which preprocessing techniques may be required?

A good EDA often contributes more to model performance than simply changing algorithms.

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 

# Display settings
pd.set_option('display.max_columns', None)


#Plot Settings
plt.style.use("default")
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("../data/ford.csv")
df.head()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df["price"],bins=40)

plt.title("Distribution of Car Prices")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df["mileage"], bins=40)

plt.title("Mileage Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="fuelType")

plt.title("Fuel Type Distribution")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="transmission")

plt.title("Transmission Distribution")

plt.show()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Matrix")

plt.show()
```

# Sprint 2 – Part 2

## Objective

The objective of this sprint is to perform advanced exploratory data analysis by investigating relationships between variables, identifying outliers, and extracting business insights.

These insights will help us decide which preprocessing techniques and feature engineering strategies should be applied before training machine learning models.

```python
plt.figure(figsize=(12,6))

sns.boxplot(data=df,x="year" ,y="price")

plt.title("Price Distribution by Manufacturing Year")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(data=df,x="mileage" , y="price" ,alpha = 0.6)

plt.title("Price vs Mileage")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df,x="fuelType",y="price")

plt.xticks(rotation=45)

plt.title("Price by Feul Type")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x="transmission" , y="price")

plt.title("Price by Transmission Type")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x="engineSize" , y="price")

plt.title("Engine Size vs Price")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.scatterplot(data=df, x="mpg" , y="price")

plt.title("MPG vs Price")

plt.show()
```

```python
numeric_cols = [
    "price",
    "mileage",
    "tax",
    "mpg",
    "engineSize"
]

plt.figure(figsize=(15,8))

for i , col in enumerate(numeric_cols):
    plt.subplot(2,3,i+1)
    sns.boxplot(y=df[col])
    plt.title(col)

plt.tight_layout()

plt.show()


```

```python
sns.pairplot(
    df[[
         "price",
            "mileage",
            "engineSize",
            "mpg",
            "year"
    ]]
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.violinplot(data=df, x="fuelType", y="price")

plt.xticks(rotation=45)

plt.show()
```

```python
top_models = df.groupby("model")["price"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))

sns.boxplot(x=top_models.index, y=top_models.values)

plt.title("Top 10 Most Expensive Ford Models")

plt.tight_layout()

plt.show()
```

```python
cheap_models = df.groupby("model")["price"].mean().sort_values().head(10)

plt.figure(figsize=(12,6))

sns.boxplot(x=cheap_models.index, y=cheap_models.values)

plt.title("Top 10 Most Cheapest Ford Models")

plt.tight_layout()

plt.show()
```

# Business Insights

1. Price generally increases with newer manufacturing years.

2. Mileage has a negative relationship with selling price.

3. Automatic transmission vehicles appear to have higher average prices.

4. Larger engine sizes generally correspond to higher prices.

5. Certain models consistently command premium prices.
